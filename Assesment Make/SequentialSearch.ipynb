# Search algorithms

Data can be stored in a collection such as a list. Here each data item is stored in a position relative to the other entries. In Python lists, these relative positions are the index values of the individual items. Since these index values are ordered, it is possible to visit them in sequence. If we wish to search this data for some item, this process gives rise to a simple searching technique, the sequential search.

Figure 1 shows how this search works. Starting at the first item in the list, we simply move from item to item, following the underlying sequential ordering until we either find what we are looking for or run out of items. If we run out of items, we have discovered that the item we were searching for was not present.

![image]('Assesment Make/Data/seqsearch.png')

In this example shown above, a sequence search for the item `77` would return the position index `4`. Whereas a search for the item `10` would return `Not Found`. While in this example the data is integers, the process is similar for strings or floats too.
